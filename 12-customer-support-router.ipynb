```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

## Challenge:

Build a LangGraph that routes customer requests to the correct department.

```python
class State(TypedDict):
    query: str
    result: str 
```

```python
def billing_agent(state):
    return {"result": "Billing department selected"}
```

```python
def technical_agent(state):
    return {"result": "Technical department selected"}
```

```python
def shipping_agent(state): 
    return {"result": "Shipping department selected"}
```

```python
def general_agent(state):
    return {"result": "General department selected"}
```

```python
billing_keywords = ["refund", "payment", "invoice"]
technical_keywords = ["password", "login", "account"] 
shipping_keywords = ["shipping", "delivery", "package"] 
```

```python
def router(state): 
    query = state["query"].lower()

    if any(k in query for k in billing_keywords):
        return "billing"
    elif any(k in query for k in technical_keywords):
        return "technical"
    elif any(k in query for k in shipping_keywords):
        return "shipping"
    else:
        return "general"
```

```python
builder = StateGraph(State)

builder.add_node("billing_agent", billing_agent) 
builder.add_node("technical_agent", technical_agent)
builder.add_node("shipping_agent", shipping_agent)
builder.add_node("general_agent", general_agent)
builder.add_node("router", lambda state:state)

builder.add_edge(START, "router")
builder.add_conditional_edges(
    "router",
    router,
    {
        "billing": "billing_agent",
        "technical": "technical_agent",
        "shipping": "shipping_agent",
        "general": "general_agent"
    }
)
builder.add_edge("billing_agent", END)
builder.add_edge("technical_agent", END)
builder.add_edge("shipping_agent", END)
builder.add_edge("general_agent", END)

builder.compile()
graph = builder.compile()
```

```python
graph.invoke(
    {"query": "I have a shipping inquiry!"}
)
```
